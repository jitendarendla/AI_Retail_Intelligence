# 📈 Module 3: Exploratory Data Analysis (EDA)

In this notebook, we calculate platform Key Performance Indicators (KPIs) and visualize sales patterns, temporal dynamics, top performing products/customers, and geographic distributions using Plotly and Matplotlib.

---

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

cleaned_path = os.path.join("..", "data", "processed", "cleaned_retail_data.csv")
df = pd.read_csv(cleaned_path)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
print(f"Loaded Cleaned Data: {df.shape[0]:,} rows")
```

### Step 1: Calculate Dashboard KPIs

```python
total_revenue = df["TotalPrice"].sum()
total_orders = df["InvoiceNo"].nunique()
total_customers = df["CustomerID"].nunique()
total_products = df["StockCode"].nunique()
avg_order_value = total_revenue / total_orders

print(f"💰 Total Revenue: £{total_revenue:,.2f}")
print(f"📦 Total Orders: {total_orders:,}")
print(f"👥 Total Customers: {total_customers:,}")
print(f"🏷️ Total Products: {total_products:,}")
print(f"💳 Average Order Value: £{avg_order_value:.2f}")
```

### Step 2: Monthly Revenue Trend Plot

```python
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)
monthly_rev = df.groupby('YearMonth')['TotalPrice'].sum().reset_index()

fig = px.line(monthly_rev, x='YearMonth', y='TotalPrice', markers=True, title='📈 Monthly Revenue Trend (£)')
fig.show()
```

### Step 3: Top 10 Products by Revenue

```python
top_prods = df.groupby('Description')['TotalPrice'].sum().sort_values(ascending=False).head(10).reset_index()

fig = px.bar(top_prods, y='Description', x='TotalPrice', orientation='h', title='🏆 Top 10 Revenue Products', color='TotalPrice')
fig.update_layout(yaxis=dict(autorange='reversed'))
fig.show()
```

### Step 4: Top 10 Customers by Spend

```python
top_cust = df.groupby('CustomerID')['TotalPrice'].sum().sort_values(ascending=False).head(10).reset_index()
top_cust['CustomerID'] = 'Cust #' + top_cust['CustomerID'].astype(str)

fig = px.bar(top_cust, x='CustomerID', y='TotalPrice', title='👑 Top 10 High-Value Customers', color='TotalPrice')
fig.show()
```

### Step 5: Country-wise Revenue Contribution

```python
country_rev = df.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False).head(10).reset_index()
fig = px.pie(country_rev, names='Country', values='TotalPrice', title='🌍 Top 10 Country Revenue Share', hole=0.4)
fig.show()
```

### Step 6: Revenue by Day of Week & Hour

```python
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Sunday']
weekday_rev = df.groupby('Weekday')['TotalPrice'].sum().reindex(days_order).reset_index()
hour_rev = df.groupby('Hour')['TotalPrice'].sum().reset_index()

fig1 = px.bar(weekday_rev, x='Weekday', y='TotalPrice', title='📅 Sales Revenue by Weekday')
fig1.show()

fig2 = px.line(hour_rev, x='Hour', y='TotalPrice', markers=True, title='⏰ Sales Revenue by Hour of Day')
fig2.show()
```

# 🔍 Module 1: Data Understanding

In this notebook, we load the raw dataset `Online Retail.xlsx` and perform initial data inspection to understand dataset shape, column datatypes, missing values, duplicate records, and summary statistics.

---

### Step 1: Import Libraries & Load Dataset

```python
import os
import pandas as pd
import numpy as np

data_path = os.path.join("..", "data", "raw", "Online Retail.xlsx")
print(f"Loading raw dataset from: {data_path}")
df_raw = pd.read_excel(data_path)
print("Dataset successfully loaded!")
```

### Step 2: Dataset Dimensions (Shape)

```python
shape = df_raw.shape
print(f"Total Rows: {shape[0]:,}")
print(f"Total Columns: {shape[1]}")
```

### Step 3: First 5 Rows Preview

```python
df_raw.head()
```

### Step 4: Data Types & Non-Null Counts

```python
df_raw.info()
```

### Step 5: Missing Values Analysis

```python
missing = df_raw.isnull().sum()
missing_pct = (missing / len(df_raw)) * 100
missing_df = pd.DataFrame({"Missing Count": missing, "Percentage (%)": missing_pct})
missing_df
```

### Step 6: Duplicate Rows Count

```python
duplicates = df_raw.duplicated().sum()
print(f"Duplicate Rows Found: {duplicates:,}")
```

### Step 7: Descriptive Statistical Summary

```python
df_raw.describe(include="all", datetime_is_numeric=True)
```

### Step 8: Unique Entity Metrics

```python
print("Unique Customers:", df_raw["CustomerID"].dropna().nunique())
print("Unique Products (StockCodes):", df_raw["StockCode"].nunique())
print("Unique Invoices:", df_raw["InvoiceNo"].nunique())
print("Unique Countries:", df_raw["Country"].nunique())
```

### Step 9: Column Purpose & Definition Table

| Column Name | Data Type | Description & Business Meaning |
|---|---|---|
| **InvoiceNo** | Object/String | 6-digit transaction ID. If starts with 'C', indicates order cancellation. |
| **StockCode** | Object/String | 5-digit unique product catalog identifier. |
| **Description** | Object/String | Commercial product name / text description. |
| **Quantity** | Integer | Quantity of product purchased per order item. |
| **InvoiceDate** | Datetime | Exact date and time when order transaction was placed. |
| **UnitPrice** | Float | Product unit price in Sterling (£). |
| **CustomerID** | Float/String | 5-digit unique customer account identifier. |
| **Country** | Object/String | Customer country of residence. |

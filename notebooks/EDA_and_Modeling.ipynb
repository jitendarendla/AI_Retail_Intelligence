```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

# Add root directory to path
sys.path.append(os.path.abspath('..'))

from utils.data_loader import load_raw_data, get_dataset_summary, clean_data
from utils.segmentation import train_customer_segmentation, calculate_rfm
from utils.market_basket import generate_market_basket_rules, recommend_products
from utils.sales_prediction import train_sales_forecaster, prepare_daily_sales
from utils.rag_assistant import get_rag_assistant
```

## 📊 Module 1: Data Understanding

```python
raw_df = load_raw_data('../data/raw/Online Retail.xlsx')
summary = get_dataset_summary(raw_df)
print('Dataset Shape:', summary['shape'])
print('Missing Values:', summary['missing_values'])
print('Duplicate Rows:', summary['duplicate_rows'])
```

## 🧹 Module 2: Data Cleaning

```python
cleaned_df = clean_data(raw_filepath='../data/raw/Online Retail.xlsx', save_processed=False)
print('Cleaned DataFrame Shape:', cleaned_df.shape)
cleaned_df.head()
```

## 📈 Module 3: Exploratory Data Analysis

```python
print('Total Revenue: £', cleaned_df['TotalPrice'].sum())
top_products = cleaned_df.groupby('Description')['TotalPrice'].sum().sort_values(ascending=False).head(10)
print('Top 10 Products by Revenue:\n', top_products)
```

## 👥 Module 4: RFM Customer Segmentation

```python
rfm_df = train_customer_segmentation(cleaned_df)
print(rfm_df['Segment'].value_counts())
```

## 🛒 Module 5 & 6: Market Basket Analysis & Product Recommendations

```python
rules_df = generate_market_basket_rules(cleaned_df)
recs = recommend_products('WHITE HANGING HEART T-LIGHT HOLDER', rules_df, top_n=5)
print('Recommendations for WHITE HANGING HEART T-LIGHT HOLDER:', recs)
```

## 📊 Module 7: Sales Prediction

```python
eval_res = train_sales_forecaster(cleaned_df)
print(f"Model Performance -> MAE: £{eval_res['mae']:.2f}, RMSE: £{eval_res['rmse']:.2f}, R2: {eval_res['r2']:.4f}")
```

## 💬 Module 8: RAG AI Assistant

```python
assistant = get_rag_assistant()
assistant.build_knowledge_base(cleaned_df, rules_df, rfm_df)
ans = assistant.query('Which products generated highest revenue?')
print(ans)
```
